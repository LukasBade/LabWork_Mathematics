Typ 1: Skizze Polynom/LGS 

n Stützpunkte -> LGS -> np.linalg.solve(A, y_punkte)

1. Achsen und Ticks übernehmen
2. x/y Koordinaten eintragen
3. x_poly = np.linspace(x_start, x_ende, 400) Limit der Kurve eingeben! 


```python
import numpy as np
import matplotlib.pyplot as plt


x_min, x_max, y_min, y_max = 0, 5, 0, 3


fig, ax = plt.subplots(figsize=(10, 10))
ax.set_aspect('equal')
ax.axis([x_min, x_max, y_min, y_max])

# *Hauptbeschriftungen der Achsen (Schrittweite aus Skizze ablesen; Endwert mit +schritt einschließen)*
ax.set_xticks(np.arange(0, 5, 1))
ax.set_yticks(np.arange(0, 3, 1))



# *Gitterlinien aktivieren*
ax.grid(which='major', alpha=0.5)
ax.grid(which='minor', alpha=0.15)


x_punkte = np.array([1, 2, 3])
y_punkte = np.array([1, 2, 2])

# Koeffizientenmatrix A aufbauen (Spalten: x^0, x^1, ..., x^(n-1)) und LGS A * a = y lösen
grad = len(x_punkte) - 1
A = np.column_stack([x_punkte**i for i in range(grad + 1)])
a = np.linalg.solve(A, y_punkte)

# *Zeichenbereich [start_x, ende_x] des Polynoms festlegen (z. B. Bildrand oder Bereich der Punkte) und Stützstellenzahl (z. B. 400) setzen*
x_poly = np.linspace(0, 5, 500)
y_poly = sum(a[i] * (x_poly**i) for i in range(grad + 1))

# *Farbe/Linienstil für das Polynom und Farbe/Marker für die Stützpunkte aus der Aufgabe übernehmen*
ax.plot(x_poly, y_poly, "b*", linewidth=1.5)
ax.plot(x_punkte, y_punkte, "r*", markersize=7)

display(fig)
```
